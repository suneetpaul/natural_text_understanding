```python
import sqlite3
import pandas as pd
import numpy as np
import os
```

```python
conn = sqlite3.connect('vector-DB.db')

cursor = conn.cursor()



```

```python
cursor.execute(
    """
    Create Table IF NOT EXISTS vectors(
      id integer primary key,
      vector blob not null
    )
    """
)
conn.commit()
```

```python
vec1 = np.array([1.2, 3.4, 2.1 , 0.8])
vec2 = np.array([2.7 , 1.5, 3.9, 2.3])
```

```python
vec1.tobytes()    # numpy to bytes stream
```

```python
cursor.execute("insert into vectors (vector) values(?)",
               (sqlite3.Binary(vec1.tobytes()),))
conn.commit()
```

```python
cursor.execute("insert into vectors (vector) values(?)",
               (sqlite3.Binary(vec2.tobytes()),))
conn.commit()
```
