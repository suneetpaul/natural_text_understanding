```python
import sqlite3
```

```python
conn = sqlite3.connect("test.db")
```

```python
cursor = conn.cursor()
```

```python
cursor.execute(
    '''
    CREATE TABLE IF NOT EXISTS test(
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT NOT NULL,
        age INTEGER NOT NULL,
        department TEXT
    )
    '''
)
```

```python
cursor.execute(
    "INSERT INTO test (name , age , department) VALUES(?,?,?)",
    ("JOHN DOE" , 30 , "SALES")
)
```

```python
cursor.execute(
    "select * from test"
)
```

```python
rows = cursor.fetchall()
```
